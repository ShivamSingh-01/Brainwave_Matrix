```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn import metrics
```

```python
gold_data = pd.read_csv('/content/gld_price_data.csv')
```

```python
correlation=gold_data.drop('Date',axis=1).corr()
```

### Convert 'Date' column to datetime objects

```python
gold_data['Date'] = pd.to_datetime(gold_data['Date'])
gold_data.info()
```

```python
#constructing a heatmap to understnd the correlation
plt.figure(figsize=(8,8))
sns.heatmap(correlation,cbar=True,square=True,fmt='.1f',annot=True, annot_kws={'size':8},cmap='Blues')
```

```python
#correlation values of GLD
print(correlation['GLD'])
```

```python
#check the distribution of the gold price
sns.distplot(gold_data['GLD'],color='green')
```

```python
#splitting the features and target
X=gold_data.drop(['Date','GLD'],axis=1)
Y=gold_data['GLD']
```

```python
print(X)
```

```python
print(Y)
```

```python
#train-test split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
regressor = RandomForestRegressor(random_state=2)
regressor.fit(X_train,Y_train)
```

```python
Y_pred = regressor.predict(X_test)
```

```python
# Re-running train-test split after updating X
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,random_state=2)
```

```python
# Fitting the regressor with the updated X_train
regressor = RandomForestRegressor(random_state=2)
regressor.fit(X_train,Y_train)
```

```python
#prediction
test_data_pred= regressor.predict(X_test)
```

```python
print(test_data_pred)
```

```python
#r squared error
error_score=metrics.r2_score(Y_test,test_data_pred)
print("r squared error:", error_score)
```

```python
Y_test=list(Y_test)
```

```python
plt.plot(Y_test,color='blue',label='Actual Value')
plt.plot(test_data_pred,color='green',label='Predicted Value')
plt.title('Actual Price vs Predicted Price')
plt.xlabel('Number of values')
plt.ylabel('GLD Price')
plt.legend()
plt.show()
```
